```python
gio = 200
mua = 300
tai_san = 10000
he_so_bdkh = 1.1

print("Tốc độ gió là:", gio)
```

```python
if gio > 150:
      P = 0.05
else:
      P = 0.20

print("Xác suất P là:", P)
```

```python
# Tính tỷ lệ thiệt hại
damage_rate = (((gio / 250)**2 * 0.4) + ((mua / 600) * 0.4)) * he_so_bdkh

# Tính thiệt hại C
C = tai_san * damage_rate

# Tính rủi ro E(Risk)
E_Risk = P * C

print("Tỷ lệ thiệt hại (%):", damage_rate)
print("Thiệt hại thực tế C (Tỷ):", C)
print("Rủi ro kỳ vọng E(Risk) (Tỷ):", E_Risk)
```

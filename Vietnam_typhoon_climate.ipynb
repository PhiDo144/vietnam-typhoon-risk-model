```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập giao diện biểu đồ và cố định ngẫu nhiên để kết quả không đổi
sns.set_theme(style="whitegrid")
np.random.seed(42)

# Khai báo số lượng bão và danh sách các tỉnh
n_events = 200
provinces = ['Quảng Ninh', 'Hải Phòng', 'Thái Bình', 'Nam Định', 'Hà Nội', 'Lạng Sơn']

print("Đã khởi tạo xong thư viện và thông số cơ bản!")
```

```python
# 1. Khai báo biến đầu vào (Ví dụ)
gio = 150            # Tốc độ gió (km/h)
mua = 300            # Lượng mưa (mm)
he_so_bdkh = 1.2      # Hệ số biến đổi khí hậu
tai_san = 500        # Giá trị tài sản (Tỷ VNĐ)
P = 0.05             # Xác suất xảy ra thiên tai (5% = 0.05)

# 2. Tính tỷ lệ thiệt hại và khống chế trần tối đa 100% (1.0)
raw_damage_rate = (((gio / 250)**2 * 0.5) + ((mua / 600) * 0.5)) * he_so_bdkh
damage_rate = min(1.0, raw_damage_rate)  # Giới hạn tối đa 100%

# 3. Tính thiệt hại thực tế C
C = tai_san * damage_rate

# 4. Tính rủi ro kỳ vọng E(Risk)
E_Risk = P * C

# 5. In kết quả chuẩn định dạng
print(f"Tỷ lệ thiệt hại: {damage_rate * 100:.2f}%")
print(f"Thiệt hại thực tế C: {C:.2f} Tỷ")
print(f"Rủi ro kỳ vọng E(Risk): {E_Risk:.4f} Tỷ")
```
